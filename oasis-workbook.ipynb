## Application: The Oasis Reunion Tour

Let's bring together everything from this session — variables, arithmetic, conditionals, and while loops — on a real example.

After years of speculation, Oasis announced their 2025 reunion tour. Over 10 million fans tried to buy tickets for the available **1.4 million seats**. To maximise revenue in the face of this overwhelming demand, Ticketmaster implemented a **dynamic pricing model**.

The company set the initial ticket price at **£148.50**. For the first million tickets sold, the price increased by **£2** after every **100,000 tickets** sold. The sale of that millionth ticket, though, triggered a **£50** price increase instead. Each subsequent 100,000 tickets sold also incurred a £50 price increase, instead of the £2 price increase.

As prices climbed, public criticism mounted. The government is now considering introducing a **30% cap** above the initial price for any ticket sold through dynamic pricing models. Ticketmaster needs to calculate how this potential price cap would have affected their revenue.

### Building up the solution

Let's build this step by step, the same way we've approached problems all session.

**Step 1: track tickets sold.** We sell tickets in blocks of 100,000, until we reach 1.4 million:

```python
# Step 1: track tickets sold
tickets_sold = 0

while tickets_sold < 1.4e6:
    tickets_sold += 1e5
    print(tickets_sold)

print(tickets_sold)
```

**Step 2: track the price.** Starting at £148.50, the price goes up by £2 per block until we pass 1 million tickets, then by £50 per block. For example, just past the 1 million mark:

```python
# Step 2: track the price
tickets_sold = 0
init_price = 148.5
price = init_price
while tickets_sold < 1.4e6:
    tickets_sold += 1e5
    if tickets_sold < 1e6:
        price += 2
    else: 
        price += 50
    print(f'tickets_sold={tickets_sold}, Price={price}')

print(tickets_sold)
```

**Step 3: put it together, and track revenue too.** Each block of 100,000 tickets sells at the *current* price, so we add `price * 100_000` to a running total each time round the loop:

```python
# Step 3: put it together, and track revenue too

total_revenue = 0
tickets_sold = 0
init_price = 148.5
price = init_price

while tickets_sold < 1.4e6: 
    tickets_sold += 1e5
    total_revenue += price * tickets_sold

    if tickets_sold < 1e6: 
        total_revenue += price * 1e5 
        curr_price = price
        price += 2
        print(f'tickets sold = {tickets_sold} at price ={curr_price} giving total revenue ={total_revenue}')
    else: 
        total_revenue += price * 1e5  
        curr_price = price
        price += 50
        print(f'tickets sold = {tickets_sold} at price ={curr_price} giving total revenue ={total_revenue}')
```
